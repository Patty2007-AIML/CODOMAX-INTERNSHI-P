```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("exam_score_predictions.csv")
sample = df.head(50)

plt.scatter(sample["id"], sample["exam_score"])
plt.xlabel("Student ID")
plt.ylabel("Exam Score")
plt.title("Student ID vs Exam Score")
plt.show()

plt.bar(sample["id"].astype(str), sample["exam_score"])
plt.xlabel("Student ID")
plt.ylabel("Exam Score")
plt.title("Exam Scores of First 50 Students")
plt.xticks(rotation=90)
plt.show()

plt.plot(sample["id"], sample["exam_score"], marker="o")
plt.xlabel("Student ID")
plt.ylabel("Exam Score")
plt.title("Exam Score Trend")
plt.show()
```
